```python
import re
def convert_to_numeric(time_string):
    """
    Extracts the numerical value from a string like "15 M",
    removes "M", and converts it to an integer.

    Args:
        time_string (str): The time string to process (e.g., "15 M").

    Returns:
        int: The extracted numerical value as an integer.
    """
    if pd.isnull(time_string):
        return None
    try:
        # Extract the numerical part using regular expression
        numeric_part = re.search(r'(\d+)', time_string)
        if numeric_part:
            numeric_time = int(numeric_part.group(1))  # Convert to integer
            return numeric_time
        else:
            return None
    except (ValueError, AttributeError):
        # Handle potential errors
        return None


# Apply the function to the 'prep_time' column
data['prep_time'] = data['prep_time'].apply(convert_to_numeric)
data['cook_time'] = data['cook_time'].apply(convert_to_numeric)
```

```python
# Define cuisine grouping
indian_cuisines = [
    'South Indian Recipes', 'North Indian Recipes', 'Indian', 'Coastal Karnataka', 'Karnataka',
    'Goan Recipes', 'Kerala Recipes', 'Hyderabadi', 'Chettinad', 'Udupi', 'Mughlai',
    'Kashmiri', 'Andhra', 'Mangalorean', 'Bengali Recipes', 'Himachal', 'Tamil Nadu',
    'Gujarati Recipes﻿', 'Punjabi', 'Konkan', 'Maharashtrian Recipes', 'Rajasthani',
    'Oriya Recipes', 'North Karnataka', 'Assamese', 'Sindhi', 'Uttarakhand-North Kumaon ',
    'North East India Recipes', 'Uttar Pradesh', 'Bihari', 'Haryana', 'South Karnataka',
    'Malabar', 'Parsi Recipes', 'Lucknowi', 'Jharkhand', 'Kongunadu'
]

asian_cuisines = [
    'Chinese', 'Thai', 'Asian', 'Japanese', 'Vietnamese', 'Indonesian',
    'Korean', 'Burmese', 'Sri Lankan', 'Nepalese', 'Hunan', 'Cantonese',
    'Sichuan', 'Singapore', 'Shandong'
]

western_cuisines = [
    'Continental', 'Italian Recipes', 'French', 'American', 'European',
    'British', 'Greek', 'Spanish'
]

middle_eastern = ['Middle Eastern', 'Arab', 'Afghan']

other_cuisines = [
    'Fusion', 'Mediterranean', 'African', 'Caribbean ', 'Jewish',
    'Pakistani', 'Bangladeshi', 'Malaysian', 'Swedish'
]

# Function to map cuisine
def map_cuisine(cuisine):
    if pd.isna(cuisine):
        return 'Unknown'
    cuisine = cuisine.strip()
    if cuisine in indian_cuisines:
        return 'Indian'
    elif cuisine in asian_cuisines:
        return 'Asian'
    elif cuisine in western_cuisines:
        return 'Western'
    elif cuisine in middle_eastern:
        return 'Middle Eastern'
    elif cuisine in other_cuisines:
        return 'Other'
    else:
        return 'Other'

# Apply mapping
data['cuisine'] = data['cuisine'].apply(map_cuisine)
```

```python
import re

def contains_hindi(text):
    if pd.isna([text]):
        return False
    return bool(re.search('[\u0900-\u097F]', text))

# Count how many rows in 'ingredients' have Hindi
hindi_ingredients_count = data['ingredients'].apply(contains_hindi).sum()
print(f"Rows with Hindi in 'ingredients': {hindi_ingredients_count}")

# Optional: for tags column
hindi_tags_count = data['tags'].apply(contains_hindi).sum()
print(f"Rows with Hindi in 'tags': {hindi_tags_count}")

hindi_description_count = data['description'].apply(contains_hindi).sum()
print(f"Rows with Hindi in 'description': {hindi_description_count}")

hindi_instructions_count = data['instructions'].apply(contains_hindi).sum()
print(f"Rows with Hindi in 'instruction': {hindi_instructions_count}")

hindi_recipe_title_count = data['recipe_title'].apply(contains_hindi).sum()
print(f"Rows with Hindi in 'recipe_title': {hindi_recipe_title_count}")
```

```python
def is_non_english_hindi(text):
    if pd.isna(text):
        return False
    # Keep rows that have none of English or Hindi characters
    return not re.search(r'[a-zA-Z\u0900-\u097F]', text)

# Filter rows from 'ingredients' that have no English or Hindi
non_eng_hindi_ingredients = data[data['ingredients'].apply(is_non_english_hindi)]

# View or count them
print("Rows with neither English nor Hindi in 'ingredients':", len(non_eng_hindi_ingredients))
print(non_eng_hindi_ingredients[['ingredients']])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer


tfidf_vectorizer = TfidfVectorizer()

# Vectorize ingredients
ingredients_vectorized = tfidf_vectorizer.fit_transform(data['ingredients'])

# Convert to dense matrix to view
ingredients_dense = ingredients_vectorized.toarray()

print("Ingredients TF-IDF :", ingredients_dense)
print("Ingredients TF-IDF shape:", ingredients_dense.shape)
```

```python
# Vectorize tags
tags_vectorized = tfidf_vectorizer.fit_transform(data['tags'])

# Convert to dense matrix to view
tags_dense = tags_vectorized.toarray()

# Print the shape to check
print("Tags TF-IDF shape:", tags_dense.shape)
```

```python
data
```

```python
# weights
w_cuisine = 0.4
w_course = 0.5
w_diet = 0.6
w_time = 0.2
w_ingredients = 6.0  # HIGHER priority
w_tags = 1.0
```

```python
from scipy.sparse import hstack
from scipy.sparse import csr_matrix
from sklearn.metrics.pairwise import cosine_similarity


# Convert the basic features to sparse matrix and apply weights
basic_features = csr_matrix(data[['cuisine', 'course', 'diet', 'overall_time']].values.astype(float))
basic_features[:, 0] *= w_cuisine
basic_features[:, 1] *= w_course
basic_features[:, 2] *= w_diet
basic_features[:, 3] *= w_time

# Apply weights to TF-IDF features
ingredients_weighted = csr_matrix(ingredients_dense * w_ingredients)
tags_weighted = csr_matrix(tags_dense * w_tags)

# Stack all weighted features
final_feature_matrix = hstack([
    basic_features,
    ingredients_weighted,
    tags_weighted
])

```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
cosine_sim = cosine_similarity(final_feature_matrix)


# Recommendation function
def recommend_recipes(index, top_n=10):
    sim_scores = list(enumerate(cosine_sim[index]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    sim_scores = sim_scores[1:top_n+1]  # skip itself
    recipe_indices = [i[0] for i in sim_scores]
    return data.iloc[recipe_indices]

```
